###fixing missing values in price

```python
df["price_usd"].median()
```

```python
df.groupby(["brand", "model"])["price_usd"].median()
```

```python
df.groupby(["brand", "model"])["price_usd"].agg(["count", "mean", "median"])
```

```python
#transforming the groupby price with median
df["price_usd"] = df["price_usd"].fillna(df.groupby(["brand", "model"])["price_usd"].transform("median"))
```

```python
#filling the overall price with median
df["price_usd"] = df["price_usd"].fillna(df["price_usd"].median())
```

```python
df.isnull().sum()/df.shape[0]*100
```

```python
#price local column is redundant and not uniform, dropping it from the df
df = df.drop(["price_local"], axis=1)
```

##Fixing Missing Values in Rating

```python
df["rating"].median()
```

```python
df.groupby(["brand", "model"])["rating"].median()
```

```python
#transforming the groupby ratings with median

df["rating"] = df["rating"].fillna(df.groupby(["brand", "model"])["rating"].transform("median"))
```

```python
#filling the overall rating with median

df["rating"] = df["rating"].fillna(df["rating"].median())
```

##Fixing Missing Values in Sentiment rating

```python
#cateogircal colum, hence fill with mode
def map_sentiment(rating):
    if rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"

mask = df["sentiment"].isnull()

df.loc[mask, "sentiment"] = df.loc[mask, "rating"].apply(map_sentiment)

print('Remaining missing values: ', df["sentiment"].isna().sum())
```

##Filling Missing values in Source

```python
df.source.mode()
```

```python
df["source"] = df["source"].fillna(df["source"].mode())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

##fixing date

```python
dash = df["review_date"].str.contains("-")
slash = df["review_date"].str.contains("/")
```

```python
df.loc[dash, "review_date"] = pd.to_datetime(df.loc[dash, "review_date"], format="%d-%m-%Y")

df.loc[slash, "review_date"] = pd.to_datetime(df.loc[slash, "review_date"], format="%m/%d/%Y")
```

```python
df["review_date"].isna().sum()
```

```python
df["review_date"] = pd.to_datetime(df["review_date"]).dt.date
```
